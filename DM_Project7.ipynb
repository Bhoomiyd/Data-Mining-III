```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("/content/Healthcare_Synthetic_Dataset (1).csv")

data.head()
```

```python
print("Dataset Shape:", data.shape)

data.info()
```

```python
data.isnull().sum()
```

```python
data = data.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", data.shape)
```

```python
data.describe()
```

```python
disease = data["Disease"].value_counts()

plt.figure(figsize=(6,5))

plt.bar(disease.index, disease.values)

plt.title("Disease Distribution")
plt.xlabel("Disease")
plt.ylabel("Number of Patients")

plt.xticks(rotation=20)

plt.show()
```

```python
gender = data["Gender"].value_counts()

plt.figure(figsize=(5,4))

plt.pie(
    gender,
    labels=gender.index,
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.hist(data["Age"], bins=15)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.hist(data["BloodPressure"], bins=20)

plt.title("Blood Pressure Distribution")
plt.xlabel("Blood Pressure")
plt.ylabel("Patients")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.hist(data["SugarLevel"], bins=20)

plt.title("Sugar Level Distribution")
plt.xlabel("Sugar Level")
plt.ylabel("Patients")

plt.show()
```

```python
avg_stay = data.groupby("Disease")["HospitalStay"].mean()

plt.figure(figsize=(6,4))

plt.bar(avg_stay.index, avg_stay.values)

plt.title("Average Hospital Stay by Disease")
plt.xlabel("Disease")
plt.ylabel("Average Days")

plt.xticks(rotation=20)

plt.show()
```

```python
recovery = data["RecoveryStatus"].value_counts()

plt.figure(figsize=(5,5))

plt.pie(
    recovery,
    labels=recovery.index,
    autopct="%1.1f%%"
)

plt.title("Recovery Status")

plt.show()
```

```python
bp = data.groupby("Disease")["BloodPressure"].mean()

plt.figure(figsize=(6,4))

plt.bar(bp.index, bp.values)

plt.title("Average Blood Pressure by Disease")
plt.xlabel("Disease")
plt.ylabel("Blood Pressure")

plt.xticks(rotation=20)

plt.show()
```

```python
sugar = data.groupby("Disease")["SugarLevel"].mean()

plt.figure(figsize=(6,4))

plt.bar(sugar.index, sugar.values)

plt.title("Average Sugar Level by Disease")
plt.xlabel("Disease")
plt.ylabel("Sugar Level")

plt.xticks(rotation=20)

plt.show()
```

```python
print("MEDICAL INSIGHTS REPORT")

print("\nTotal Patients:", len(data))

print("\nAverage Age:", round(data["Age"].mean(),2))

print("\nAverage Blood Pressure:", round(data["BloodPressure"].mean(),2))

print("\nAverage Sugar Level:", round(data["SugarLevel"].mean(),2))

print("\nAverage Hospital Stay:", round(data["HospitalStay"].mean(),2), "days")

print("\nMost Common Disease:")
print(data["Disease"].value_counts().head(1))

print("\nRecovery Status:")
print(data["RecoveryStatus"].value_counts())
```
